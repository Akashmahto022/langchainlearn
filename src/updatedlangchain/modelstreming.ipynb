```python
## GROQ MODEL INTEGRATION

import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:openai/gpt-oss-120b")

reponse = model.invoke("What is the capital of France?")
reponse
```

```python
for chunk in model.stream("write 200 words article on ai"):
    print(chunk.text, end="|", flush=True)
```

```python
reponse = model.batch(
    [
        "why humans talk to each other?",
        "write 200 words article on ai",
        "how can i learn langchain?",
    ]
)

for responses in reponse:
    for response in responses:
        print(response)
```

```python
model.batch(
    [
        "why humans talk to each other?",
        "write 200 words article on ai",
        "how can i learn langchain?",
    ],
    config={
        'max-concurrency': 5 # limit the number of concurrent requests to 5
    }
)
```
